```python
# Loading the dataset using pandas
import pandas as pd

df = pd.read_csv(r"D:\Babes office data\DATA ANALYSIS PROJECT 1\customer_shopping_behavior.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
# Summary statistics using .describe()

df.describe(include = 'all')
```

```python
# checking for missing data

df.isnull().sum()
```

```python
# Imputing the missing values of the review rating column with the median of each product category

df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# Renaming columns according to snake casing for better readability and documentation


df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')

df = df.rename(columns = {'purchase_amount_(usd)':('purchase_amount')}, )
```

```python
df.columns
```

```python
# create a new column 'Age group'

labels = ['Young Adult',
          'Adult',
          'Middle-Aged',
          'Senior']

df['age_group'] = pd.qcut(df['age'],q=4, labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# create a new column purchase frequency days

frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Annually' : 365,
    'Bi-Weekly': 14,
    'Every 3 Months': 90

}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping).astype(int)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df['purchase_frequency_days'].isna().sum()
```

```python
# comparing 'discount_applied' and 'promo_code_used' columns
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
# Dropping 'promo_code_used' column as it is redundant
df = df.drop('promo_code_used', axis=1)
```
